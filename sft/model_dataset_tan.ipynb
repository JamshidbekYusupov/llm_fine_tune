```python
import json
from pathlib import Path

path = Path(r"C:/Post Training/sft/hug_dataset/raw/sft_uz/output_uz_sft.json")
# C:\Post Training\sft\hug_dataset\raw\sft_uz\output_uz_sft.json
rows = json.loads(path.read_text(encoding="utf-8"))

def as_text(v):
    if isinstance(v, list):
        v = v[0] if v else ""
    return v if isinstance(v, str) else ""

q = sorted(len(as_text(r.get("prompt"))) for r in rows)
a = sorted(len(as_text(r.get("answer"))) for r in rows)
for name, xs in (("savol", q), ("javob", a)):
    pct = {p: xs[int(len(xs) * p / 100) - 1] for p in (25, 50, 75, 90, 95, 99)}
    print(name, "belgi:", pct)
```
